```python
a = int(input("Valor de 'a': "))

if a == 0:
    print("Não é uma equação de 2° grau.")
else:
    b = int(input("Valor de 'b': "))
    c = int(input("Valor de 'c': "))

    delta = b**2 - 4*a*c

    if delta < 0:
        print("Não existem raízes.")
    elif delta == 0:
        print("Existe apenas uma raíz que é", (-b + delta**0.5) / (2*a))
    else:
        print("Existem duas raízes, e são elas:")

raiz1 = (-b + delta**0.5) / (2*a)
raiz2 = (-b - delta**0.5) / (2*a)

print(f"{raiz1:.2f}")
print(f"{raiz2:.2f}")
```

```python
from random import randint

# Leitura do documento
documento = open('Forca.txt', 'r')
jogo = []
for i in documento:
    jogo.append(i.rstrip())

# Escolha de uma palavra aleatória
final = len(jogo) - 1
palavra = jogo[randint(0, final)]
print(palavra)

# Criação da sequência de underlines
underline = []
for i in range(len(palavra)):
    underline.append('_')

print(' '.join(underline))

# Tentativas do Player
while '_' in underline:
    letra = input('Digite letra: ')

    if letra in palavra:
        for i in range(len(palavra)):
            if palavra[i] == letra:
                underline[i] = letra

    print(' '.join(underline))
```

```python
matriz = [['-','-','-'],
          ['-','-','-'],
          ['-','-','-']]

# Mostra tabuleiro inicial
for linha in matriz:
    print(' '.join(linha))
print()

for jogada in range(9):
    # Alterna entre 'x' e 'o'
    marca = 'xo'[jogada % 2]

    ok = False
    while not ok:
        posicao = input("Selecione a posição (linha,coluna): ")

        try:
            linha = int(posicao.split(',')[0]) - 1
            coluna = int(posicao.split(',')[1]) - 1

            # Verifica se a posição existe
            if 0 <= linha < 3 and 0 <= coluna < 3:

                # Verifica se está livre
                if matriz[linha][coluna] == '-':
                    matriz[linha][coluna] = marca
                    ok = True
                else:
                    print("Posição já ocupada!\n")
            else:
                print("Posição não existe!\n")

        except:
            print("Entrada inválida! Use formato linha,coluna (ex: 1,2)\n")

    # Mostra tabuleiro a cada jogada
    for l in matriz:
        print(' '.join(l))
    print()
```
